```python
# Attributes

# 1) ndim
# 2) shape
# 3) size
# 4) itemsize
# 5) dtype

a= np.arange(10)
b= np.arange(12, dtype= float).reshape(3,4)
c= np.arange(8).reshape(2,2,2)

# b.ndim

# 2) Shape

# a.shape
# b.shape
# c.shape


# 3) Size
# a.size
# b.size


# 4) itemsize

# a.itemsize
# b.itemsize
# c.itemsize

print(a.dtype)
print(c.dtype)
print(b.dtype)
```

# Array Opertions

```python
a = np.arange(12).reshape(3,4)
b= np.arange(12,24).reshape(3,4)

a
```

```python
# 1) Scalar operation
# means operate with the single number like (a * 3) or (a + 4)

# a **3 
# a+4

# 2) Vetor operatoin

a**b
```

```python
# relational operators

# b > 5
b==13

b<= 34
```

```python
c = a**b
# print(c)

# print(np.max(c))
# print(np.min(c))

# to get the max and min on the axis, so the code will be 

# 0 ----> column and 1 -----> Row

np.max(c, axis = 0)
np.max(a, axis = 1)

np.prod(a, axis = 0)   # to get the product
```

```python
# mean / median / std / var
np.mean(b, axis = 1)
```

```python
np.median(a, axis = 0)
```

```python
np.std(a, axis =1)
np.std(a)
```

```python
np.var(b)
np.var(b, axis = 1)
```

```python
# Dot product  --> goes with the same matrix like --- (3,4).(4,3)  ---> res- (3,3)
# or (12,4).(4,12)  and result will be (12, 12)

# ex

a1 = np.arange(12).reshape(3,4)
a2 = np.arange(12).reshape(4,3)

np.dot(a1,a2)
```

```python
# log and exponents function

np.log(a1)

np.exp(a1)
```

```python
# round/ Floor/ Ceil functions

np.round
np.random.random((2,3))*100  # *100 for scalr values 
# np.round - round of the value at the nearest integer value, ex 5.6 > covert it to 6

np.round(np.random.random((2,3))*100 )

# np.floor - gives us the  round of the lesser nearest values, ex - 6.9 > convert it to 6

np.ceil(np.random.random((2,3))*100)  # incresing nearest value to the integer 
```
