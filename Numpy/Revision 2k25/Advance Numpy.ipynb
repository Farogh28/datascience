# Broadcasting

```python
"""The term broadcasting describes how NumPy treats arrays with different 
shapes during arithmetic operations.
the smaller array is “broadcast” across the larger array so that they have compatible shapes"""
```

```python
a = np.arange(6).reshape(2,3)
b = np.arange(3).reshape(1,3) 

print (a)
print (b)
print ()

print (a+b)
```

```python
# Rules -- 1) Arrays must be in the same dimension (means, broadcast automatically strech the smaller matrix to the comparable size)
2)# if dimension are same then it will not change
```

```python
a = np.arange(3).reshape(1,3)
b = np.arange(3).reshape(3,1) 
print (a)

print (b)
print()
print (a + b)  # Broadcasting
```

```python
a = np.arange(12).reshape(3,4)  
b = np.arange(12).reshape(4,3)   # can't broadcast or can't strech 3 upto 4. coz we strech the one only

print(a)

print(b)
print()

print(a+b)
```

```python
# if dimension are same then it will not change, broadcast can't strech
#if dim is 1 D so brodcst strech to 2 D/ 3D by adding 1 

a = np.arange(16).reshape(4,4)
b = np.arange(4).reshape(2,2)

print (a)
print (b)

print()
print (a+b)
```

```python
a = np.arange(8).reshape(2,4)
b = np.arange(4).reshape(4)
print (a)
print()
print(b)
print()

print(a+b)
```

# Mean Squared Error  (LOSS Function)

```python
actual = np.random.randint(1,50,25)
predicted = np.random.randint(1,50,25)
```

```python
def mse (actual, predicted):
    return np.mean((actual - predicted)**2)  # Fourmula

mse(actual, predicted)
```
