```python
a = np.arange(10)
b = np.arange(12).reshape(3,4)  # (3,4) means 3 rows and 4 columns
c = np.arange(8).reshape(2,2,2)
d = np.arange(16).reshape(2,2,2,2)

# print(d)
```

# Indexing

```python
# --------For 2D array indexing------------
print(b)

print ("   ")

# to get particular element in 2D array
#syntax ---->  b[ row no. , column no] 

# searching for 6 
b[1, 2]

# serching to 11
b[2,3]
```

```python
# For 3D Tensor

print(c)

# search element 5

# Syntax ----   c[array no, row no, columns no]

c[1,0,1]
```

```python
# For 4 D matrix
print(d)

# syntax - d[array no(3d), array no(2d), row no., column no]

# searching for element 13
# d[1,1,0,1]

# Searching for element 7
d[0,1,1,1]
```

# Slicing

```python
# for 1D 

print (a)

# to get 2,3,4

a[2:5] #(start to n-1)

a[6:] # to get the element up to the last

a[2:7:2] #(start , end (n-1), gap/skip)
```

```python
# for 2D 
print(b)

# b[row no. , column no]
print ("  ")

# getting only 1st row
b[0,:]

# get [[5 6], [9 10]] 
# [row 1 :(onwards), column no 1 to 3(n-1) ]
b[1:, 1:3]

# get the[[2 3],[6 7],[10,11]]
# b[row 0 : (all), from column 2: (all)]
b[0: , 2:]


# get 0  3  8  11 ( all the corner no.)
# b[:(all rows), ]
b[::2,::3]

# get 1 3 9 11 
b[::2,1::2 ]

# get 4 and 7
# b[row 1, colomn no1 and skip 3 and last ]
b[1, 0::3]

# get [[1 2 3], [5 6 7]]
b[0:2, 1:]
```

```python
# for the center array 
c[1]

# for the 1st and last array
c[::2]

# to get the 3 4 5
# c[0 th array: first row:  all columns]
c[0,1,:]

# get 12 13 14
c[1,1,:]

# get 10 13 16
# c[1 postion array , all rows : 1st postion column]
c[1, :, 1]

# get [[22 23], [25 26]]
# c[array no.2 , row from 1:, column from 1: ]
c[2, 1:, 1:]


# get [[0 2],[18 20]]

c[::2,0,  ::2]
```
