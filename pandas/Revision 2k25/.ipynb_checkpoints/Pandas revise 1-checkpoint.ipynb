# Pandas

```python
""" pandas -is a powerful and open-source Python library. 
The Pandas library is used for data manipulation and analysis. 

Pandas consist of data structures and functions to perform efficient operations on data."""
```

```python
# PANDAS SERIES -  is like a column in a table. it is 1-D array holding data of any type.
```

```python
import pandas as pd
import numpy as np
```

```python
country = ["Ind", "Pak", "Aus", "USA", "SAF", "RUS", "SRL"]

pd.Series(country)  # Make a series
```

```python
runs = [12,2,44,23,155,20,56]
pd.Series(runs)
```

```python
# for custom index

marks = [50,69,73,79,88]
subject = ['math', 'science', 'physics', 'chemistry', 'geo']

pd.Series(marks, index=subject)
```

```python
# setting a name

marks = [50,69,73,79,88]
subject = ['math', 'science', 'physics', 'chemistry', 'geo']

marks = pd.Series(marks, index=subject, name='Received marks')
marks
```

```python
marks ={
    'math':60,
    'science': 70,
    'physics': 69,
    'geography': 90,
    'chemistry':78,  
}

marks_series= pd.Series(marks, name= 'updated mark')
marks_series
```
