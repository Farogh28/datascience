# to Fetch ROWS from the dataframe

```python
# iloc - Searches using index positions
# loc - Searches using index labels
```

```python
# for single row  - as we are taking out a single row so its type is a series
type(movies.iloc[0])
movies.iloc[5]
```

```python
type(movies.iloc[0:5])
```

```python
# multiple row

# synatx  - df_name.iloc[start : end : with gap of]
movies.iloc[5:16:2]
```

```python
movies.iloc[:3]
```

```python
# fancy indexing
# in this it returns the row which we require

movies.iloc[[0,4,5]]
```

```python
# loc  - is an instance
# it fetches the data as per the label of the index

# students.loc['kevin'] - this will work with via name if we set name column as a inex column
students.loc[3]
# students
students.loc[:4]
```

```python
students.loc[0:4:2]  #.loc[start:end: gap]
```

```python
# Selecting both rows and columns
```

```python
# wants 3 rows and 3 coloumns data  

# With the help of iloc 

"""in iloc we need to specify index positions"""

movies.iloc[0:3, 0:3].values #.values is used to convert a DataFrame or Series into a NumPy array.
```

```python
# or 

# with the help of "LOC" instance 

"""in loc instance we can give the name of the column"""


movies.loc[0:2,'title_x': 'poster_path']
```
